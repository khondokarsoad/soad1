```python
states = ["Arizona", "California", "Colorado", "Florida", "Minnesota"]
print(states)
```

```python
#list constructor
new_states = list(("Louisiana", "Maryland", "Nevada", "Montana"))
print(type(new_states))
```

```python
print(len(states))
```

```python
del states[2]
print(states)
```

```python
print("Arizona" in states)
print("Louisiana" not in states)
```

```python
#nested list
all_states = [states, new_states]
print(all_states)
```

```python
print(all_states[1][2])
```

```python
#unpacking values from a list
a, b, c, d = states
print(d)
```

```python
#slicing list
print(states[0:2])
```

```python
#adding.insert
states.insert(2,"Texas")
print(states)
```

```python
#adding.append
states.append("Colorado")
print(states)
```

```python
#adding.extend
more = ["Ohio", "Alaska"]
states.extend(more)
print(states)
```

```python
#"+" operator
more_states = states + more
print(more_states)
```

```python
#remove.remove
states.remove("Colorado")
print(states)
```

```python
#remove.pop
states.pop(-5)
print(states)
```

```python
#remove.del
del more_states[-4:-1]
print(more_states)
```

```python
#remove.clear
more_states.clear()
print(more_states)
```

```python
states.sort()
print(states)
```

```python
states.sort(reverse = True )
print(states)
```

```python
copystates = states.copy()
print(copystates is states)
print(copystates == states)
```
